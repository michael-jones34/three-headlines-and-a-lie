```python
from firecrawl import Firecrawl
from IPython.display import Markdown, display

client = Firecrawl(api_key=api_key)

result = client.scrape(
    "https://www.morningbrew.com/issues/macy-s-vs-amazon",
    formats=["markdown"]
)

display(Markdown(result.markdown))
```

```python
type(result)
```

```python
# Find the index for where the quiz section starts
quiz_section_start_index = result.markdown.find("Three Headlines and a Lie")
```

```python
# Extract from the quiz start to the end of the newsletter
quiz_section_start = result.markdown[quiz_section_start_index:]
```

```python
# Split the text up by <br> which separates out what will contain the individual questions and the answer
parts = quiz_section_start.split("<br>")
print(parts)
```

```python
print([(i, item[:100]) for i, item in enumerate(parts)])
```

```python
# Define what gives a section that wants to be kept so that all others can be removed
seeking = ["1.", "2.", "3.", "4.", "answer"]
# for item in seeking:
#     print(item in parts[2])

any(term in parts[2] for term in seeking)
```

```python
# Remove unwanted parts
[item for item in parts if any(term in item.lower() for term in seeking)]
```
