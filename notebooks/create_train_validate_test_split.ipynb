# Create the splits for the different models

```python
import json
from IPython.display import Markdown, display
import numpy
import pandas as pd
from collections import Counter

data_path = "../data"

with open(f"{data_path}/ml_dataset.json", "r") as f:
    ml_dataset = json.load(f)
```

```python
train_classic_ml = []
train_nn = []
validate = []
test = []

for index, item in enumerate(ml_dataset, start=1):
    if index % 5 == 0:
        test.append(item)
    elif index % 5 == 4:
        validate.append(item)
        train_classic_ml.append(item)
    else:
        train_classic_ml.append(item)
        train_nn.append(item)

with open(f"{data_path}/test_set.json", "w") as f:
    json.dump(test, f, indent=2)

with open(f"{data_path}/train_classic_ml_set.json", "w") as f:
    json.dump(train_classic_ml, f, indent=2)

with open(f"{data_path}/validate_set.json", "w") as f:
    json.dump(validate, f, indent=2)

with open(f"{data_path}/train_nn_set.json", "w") as f:
    json.dump(train_nn, f, indent=2)
```

```python
with open(f"{data_path}/test_set.json", "r") as f:
    test_set = json.load(f)

with open(f"{data_path}/validate_set.json", "r") as f:
    validate_set = json.load(f)

with open(f"{data_path}/train_nn_set.json", "r") as f:
    train_nn_set = json.load(f)

with open(f"{data_path}/train_classic_ml_set.json", "r") as f:
    train_classic_ml_set = json.load(f)
```

```python
df = pd.DataFrame(
    [{"date": quiz["date"], "answer": quiz["answer"]} for quiz in train_classic_ml_set]
)
df["date"] = pd.to_datetime(df["date"])
df.groupby([df["date"].dt.year, "answer"]).size().groupby(level=0).transform(lambda x: x / x.sum())
```
