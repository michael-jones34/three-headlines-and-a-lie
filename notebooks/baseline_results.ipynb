# Establish baseline results against which ML model performance can be compared

```python
import random
import json
from collections import Counter

data_path = "../data"

with open(f"{data_path}/test_set.json", "r") as f:
    test_set = json.load(f)
```

```python
test_set[0]["answer"]
```

```python
random.choice(range(0,4))
```

### Uniform random baseline

```python
accuracies = []

for iteration in range(0,100):
    outcomes = []
    for quiz in test_set:
        answer = quiz["answer"]
        guess = random.choice(range(0,4))
        if answer == guess:
            outcomes.append(1)
        else:
            outcomes.append(0)
    accuracies.append(Counter(outcomes)[1]/len(test_set))
```

```python
sum(accuracies) / len(accuracies)
```

### Majority-position baseline

```python
with open(f"{data_path}/train_classic_ml_set.json", "r") as f:
    train_classic_ml_set = json.load(f)

train_classic_ml_answers = [quiz["answer"] for quiz in train_classic_ml_set]

Counter(train_classic_ml_answers)
```

Answer index 3 has the highest count so that is the index that will be used for the majority-position baseline guess

```python
outcomes = []
for quiz in test_set:
    answer = quiz["answer"]
    if answer == 3:
        outcomes.append(1)
    else:
        outcomes.append(0)

accuracy = (Counter(outcomes)[1]/len(test_set))
accuracy
```
