Cutting everything before the quiz

```python
# Find the index for where the quiz section starts
quiz_section_start_index = result.markdown.lower().find("three headlines and a lie")
quiz_section_start_index
```

```python
# Extract from the quiz start to the end of the newsletter
quiz_section = result.markdown[quiz_section_start_index:]
quiz_section
```

Splitting into parts

```python
# Split the text up by <br> which separates out what will contain the individual questions and the answer
parts = quiz_section.split("<br>")
# print(parts)
parts
```

```python
print([(i, item[:100]) for i, item in enumerate(parts)])
```

First filter removing most of the parts

```python
# Define what gives a section that wants to be kept so that all others can be removed

import re

seeking = ["1. ", "2. ", "3. ", "4. "]

any(term in parts[2] for term in seeking)
```

```python
parts[11]
```

```python
# Looking at a regex for "Answer"

import re

matches = re.compile(r'\banswer\b')

print(matches.search(parts[11].lower()))
```

```python
# Remove unwanted parts

import re

answer_search = re.compile(r'\banswer\b')
answers_search = re.compile(r'\banswers\b')

questions_and_answers = [item for item in parts if any(term in item.lower() for term in seeking) or answer_search.search(item.lower()) or answers_search.search(item.lower())]
questions_and_answers
```

Clean question 4. and answer parts

```python
# Clean up question 4.
questions_and_answers[3] = (questions_and_answers[3].split("|"))[0]
questions_and_answers[3]
```

```python
# Find the index of the answer in the answer part
answer_index = questions_and_answers[4].lower().find("answer")
answer_index
```

```python
# Cut everything before "Answer"
answer_clean1 = questions_and_answers[4][answer_index:]
answer_clean1
```

```python
answer_parts = answer_clean1.split("|")
answer_parts
```

```python
answer_parts_stripped = [item.strip().strip("\n-") for item in answer_parts]
answer_parts_stripped
```

```python
answer_parts_stripped[0] = ""
answer_list = []

# for item in answer_parts_stripped:
#     if item:
#         answer_list.append(item)

[answer_list.append(item) for item in answer_parts_stripped if item]

answer = answer_list[0]
answer
```

```python
questions_and_answers[4] = answer
questions_and_answers
```

Separating out url from headline text

```python
# Extract URL using regex

import re

url_re = re.compile(r'(\(.*\))')
url = url_re.search(questions_and_answers[0]).group().strip("()")
url
```

```python
# Remove URL

headline = questions_and_answers[0].replace(url, "").replace("()", "").replace("[", "").replace("]", "")
```

```python
# Extract headline using regex
print(headline)
headline_re = re.compile(r'(“.*”)')
headline = headline_re.search(headline).group().strip("“”")
print(headline)
```

JSON construction

```python
quiz = {
    "issue date": "20XX-XX-XX",
    "source_url": "https://www.morningbrew.com/issues/2019%2F11%2F16",
    "headlines": [],
    "answer": answer
}
```

```python
# Function to extract URL and headline text for each item

import re

def url_headline_extract(index):
    url_re = re.compile(r'(\(.*\))')
    url = url_re.search(questions_and_answers[index]).group().strip("()")

    headline = questions_and_answers[index].replace(url, "").replace("()", "").replace("[", "").replace("]", "")
    headline_re = re.compile(r'(“.*”)')
    headline = headline_re.search(headline).group().strip("“”")

    return headline, url

```

```python
for index, item in enumerate(questions_and_answers):
    if item[0].isalpha():
        continue
    else:
        headline, url = url_headline_extract(index)
        quiz["headlines"].append({"text": headline, "url": url})

quiz
```
