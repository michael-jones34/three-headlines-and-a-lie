# Create a dataset which will be used for the quiz questions in the app

```python
import json
import re
import requests
from IPython.display import Markdown, display
import numpy
import pandas as pd

data_path = "../data"

app_quiz_path = "../web/public/data"

model_results_path = "../web/src/data"

with open(f"{data_path}/quizzes.json", "r") as f:
    quizzes = json.load(f)

with open(f"{data_path}/ml_dataset.json", "r") as f:
    ml_dataset = json.load(f)

with open(f"{data_path}/test_set.json", "r") as f:
    test_set = json.load(f)
```

```python
test_quizzes = []

for quiz in quizzes:
    if any(quiz["date"] == item["date"] for item in test_set):
        test_quizzes.append(quiz)

for quiz in test_quizzes:
    date = quiz["date"]
    ml_dataset_quiz = [item for item in ml_dataset if date == item["date"]]
    quiz["answer"] = ml_dataset_quiz[0]["answer"]

with open(f"{app_quiz_path}/app_quizzes.json", "w") as f:
    json.dump(test_quizzes, f, indent=2)
```
