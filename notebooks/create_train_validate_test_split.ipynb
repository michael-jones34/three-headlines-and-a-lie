# Create the splits for the different models

```python
import json
from IPython.display import Markdown, display
import numpy
import pandas as pd
from collections import Counter

data_path = "../data"

with open(f"{data_path}/ml_dataset.json", "r") as f:
    ml_dataset = json.load(f)
```

```python
train_classic_ml = []
train_nn = []
validate = []
test = []

for index, item in enumerate(ml_dataset, start=1):
    if index % 5 == 0:
        test.append(item)
    elif index % 5 == 4:
        validate.append(item)
        train_classic_ml.append(item)
    else:
        train_classic_ml.append(item)
        train_nn.append(item)

with open(f"{data_path}/test.json", "w") as f:
    json.dump(test, f, indent=2)

with open(f"{data_path}/train_classic_ml.json", "w") as f:
    json.dump(train_classic_ml, f, indent=2)

with open(f"{data_path}/validate.json", "w") as f:
    json.dump(validate, f, indent=2)

with open(f"{data_path}/train_nn.json", "w") as f:
    json.dump(train_nn, f, indent=2)
```
