# EDA for the dataset

```python
import json
from IPython.display import Markdown, display
import numpy
import pandas as pd
from collections import Counter

data_path = "../data"

with open(f"{data_path}/ml_dataset.json", "r") as f:
    ml_dataset = json.load(f)
```

Total number of each answer index

```python
answers = [quiz["answer"] for quiz in ml_dataset]
answer_counts = Counter(answers)
answer_counts
```

Total number of answer index by year

```python
df = pd.DataFrame(
    [{"date": quiz["date"], "answer": quiz["answer"]} for quiz in ml_dataset]
)
df["date"] = pd.to_datetime(df["date"])
df.groupby([df["date"].dt.year, "answer"]).size()
```

Proportion of answer index by year

```python
df.groupby([df["date"].dt.year, "answer"]).size().groupby(level=0).transform(lambda x: x / x.sum())
```
